### 3. Query Lebih Kompleks Menggunakan Aggregation
- **Tugas 3**: Terapkan aggregation untuk menghitung rata-rata gaji per departemen.

### 4. Latihan Tambahan
- **Latihan 4.1**: Buatlah koleksi baru `products` dan masukkan data produk (minimal 10 produk).
- **Latihan 4.2**: Lakukan query untuk menemukan produk dengan harga di atas rata-rata.
- **Latihan 4.3**: Buatlah aggregation pipeline untuk menghitung total produk dalam setiap kategori.


```python
# Latihan 4.1 : Buatlah koleksi baru `products` dan masukkan data produk (minimal 10 produk).

produk = db['products']

# Data produk
products_data = [
    {"name": "Laptop", "category": "Electronics", "price": 15000000, "stock": 15},
    {"name": "Smartphone", "category": "Electronics", "price": 8000000, "stock": 30},
    {"name": "Headphones", "category": "Accessories", "price": 500000, "stock": 50},
    {"name": "TV", "category": "Electronics", "price": 10000000, "stock": 10},
    {"name": "Blender", "category": "Home Appliances", "price": 300000, "stock": 25},
    {"name": "Washing Machine", "category": "Home Appliances", "price": 2500000, "stock": 8},
    {"name": "Microwave", "category": "Home Appliances", "price": 1200000, "stock": 12},
    {"name": "Air Conditioner", "category": "Home Appliances", "price": 4500000, "stock": 5},
    {"name": "Refrigerator", "category": "Home Appliances", "price": 5000000, "stock": 7},
    {"name": "Tablet", "category": "Electronics", "price": 4000000, "stock": 20}
]

# Masukkan data ke koleksi
produk.insert_many(products_data)

print("Data produk berhasil dimasukkan ke dalam koleksi 'products'")

# Menampilkan data dari koleksi
for product in produk.find():
    print(product)
```

```python
# Latihan 4.2 : Lakukan query untuk menemukan produk dengan harga di atas rata-rata.
rata2harga = produk.aggregate([
    {"$group": {"_id": None, "averagePrice": {"$avg": "$price"}}}
]).next()['averagePrice']

print(f"Harga rata-rata produk: {rata2harga}")

# Query produk dengan harga di atas rata-rata
diatasrata2 = collection.find({"price": {"$gt": rata2harga}})
print("Produk dengan harga di atas rata-rata:")
for product in diatasrata2:
    print(product)
```

```python
# Latihan 4.3 : Buatlah aggregation pipeline untuk menghitung total produk dalam setiap kategori.

pipeline = [
    {'$group': {'_id': '$category','total_produk': {'$sum': 1}}}
]

# Menjalankan pipeline dan mencetak hasil
for result in produk.aggregate(pipeline):
    print(result)

```

### 5. Tugas
- **Tugas 1**: Cari 5 karyawan dengan gaji tertinggi dalam setiap departemen, gunakan query atau aggregation yang sesuai.
- **Tugas 2**: Buatlah skenario di mana Anda harus menghapus karyawan yang berusia di bawah 25 tahun dari database.
- **Tugas 3**: Buatlah laporan ringkas (menggunakan MongoDB query) yang menghitung total gaji karyawan di setiap departemen, serta rata-rata umur karyawan.


```python
# Tugas 1: Cari 5 karyawan dengan gaji tertinggi dalam setiap departemen
pipeline = [
    {"$sort": {"department": 1, "salary": -1}},
    {"$group": {
        "_id": "$department",
        "topEmployees": {"$push": {"name": "$name", "salary": "$salary"}}
    }},
    {"$project": {
        "_id": 1,
        "topEmployees": {"$slice": ["$topEmployees", 5]} 
    }}
]

result = collection.aggregate(pipeline)
print("5 karyawan dengan gaji tertinggi dalam setiap departemen:")
for dept in result:
    print(f"Departemen: {dept['_id']}")
    for emp in dept['topEmployees']:
        print(f"  Nama: {emp['name']}, Gaji: {emp['salary']}")
```

```python
# Tugas 2: Hapus karyawan dengan usia di bawah 25 tahun
delete_result = collection.delete_many({"age": {"$lt": 25}})
print(f"Jumlah karyawan dihapus: {delete_result.deleted_count}")
```

```python
# Tugas 3: Hitung total gaji dan rata-rata umur dalam setiap departemen
pipeline = [
    {"$group": {
        "_id": "$department"
        "totalSalary": {"$sum": "$salary"},
        "averageAge": {"$avg": "$age"}
    }},
    {"$sort": {"_id": 1}}
]

result = collection.aggregate(pipeline)
print("Total gaji dan rata-rata umur dalam setiap departemen:")
for dept in result:
    print(f"Departemen: {dept['_id']}")
    print(f"  Total Gaji: {dept['totalSalary']}")
    print(f"  Rata-rata Umur: {dept['averageAge']:.2f}")
```
